```python
string="Pappu"
lists=[]
for letter in string:
    lists.append(letter)
print(lists)
```

```python
lists=[letter for letter in string]
print(lists)
```

```python
s=[ num for num in range(0,11)]
print(s)
```

```python
s=[ num**2 for num in range(0,11)]
print(s)
```

```python
s=[x for x in range(0,11) if x%2==0]
print(s)
```

```python
c=[0,10,20,34.5]
f=[(9/5)*t+32 for t in c]
print(f)
```

```python
f=[]
for temp in c:
    f.append(((9/5)*temp +32))
print(f)
```

```python
lists=[]
for x in[2,4,6]:
    for y in [100,200,300]:
        lists.append(x*y)
print(lists)
```

```python
lists=[x*y for x in[2,4,6] for y in [100,200,300]]
print(lists)
```

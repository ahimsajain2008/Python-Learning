```python
#ZIP FUNCTION!!
```

```python
l1=["a","b","c","d"]
l2=[1,2,3,4]
for i in zip(l1,l2):
    print(i)
```

```python
l1=["a","b","c",]
l2=[1,2,3,4]
for i in zip(l1,l2):
    print(i)
```

```python
l1=["a","b","c",]
l2=[1,2,]
for i in zip(l1,l2):
    print(i)
```

```python
l1=["a","b","c","d"]
l2=[1,2,3,4]
l3=[100,200,300,400]
for i in zip(l1,l2,l3):
    print(i)
```

```python
list(zip(l1,l2,l3))
```

```python
for a,b,c in zip(l1,l2,l3):
    print(b)
```

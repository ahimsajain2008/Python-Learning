```python
def square(num):
    return num**2
```

```python
n=[1,2,3,4,5]
```

```python
map(square,n)    #Gives memory location
```

```python
for item in map(square,n):
    print(item)         #map applies the function(square) to every element in the list (n)
```

```python
list(map(square,n)) #to get the list of numbers back
```

```python
def splicer(s):
    if len(s)%2==0:
        return 'Even'
    else:
        return s[0]
```

```python
p=['Andy', 'Eve', 'Sally']
```

```python
list(map(splicer,p))
#map(splicer, p) loops through p and passes each element into splicer for you. 
#That's why you only pass the name of the function, and map handles the execution.
```

```python
def even(n):
    return n%2==0
```

```python
n=[1,2,3,4,5,6]
```

```python
filter(even,n)#filters based on functions condition
```

```python
for n in filter(even,n):
    print(n)
```

```python
""" 
filter() checks every single item in your list.

Here is exactly how it decides what to keep and what to drop:

Passes every item: It loops through the iterable item by item and passes each one into your function.

Evaluates a Boolean condition: Your function must return True or False for each item.

Keeps only if True:

If the function returns True, that item is kept in the result.

If the function returns False, that item is thrown away (filtered out).
"""
```

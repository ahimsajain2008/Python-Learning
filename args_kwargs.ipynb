```python
def myfunc(a,b):
    #Returns 5% of the sum a and b
    return sum((a,b))*0.05
myfunc(40,60)
```

```python
def myfunc(a,b,c=0,d=0,e=0):
    #Returns 5% of the sum a and b
    return sum((a,b,c,d))*0.05
myfunc(40,60,100,100,3)
#if i enter one more value in the function call it gives me error and i want to add more values 
#then we use *args
```

```python
def myfunc(*args):
    return sum(args)*0.05
myfunc(100,2,3,40,59,78)
#Now I can enter as many numbers i can in function call
```

```python
def myfun(*args):
    print(args)#returns tuple
myfun(40,30,20,30)
```

```python
def myfunc(*args):
    for i in args:
        print(i)
myfunc(1,23,345,4567,987,78)
```

```python
def myfunc(**kwargs):
    if 'fruit' in kwargs:
        print("My fruit of choice is {}".format(kwargs['fruit']))
    else:
        print("No fruit found")
myfunc(fruit='apple')
```

```python
def myfunc(**kwargs):
    print(kwargs)#Returns Dictionary
    if 'fruit' in kwargs:
        print("My fruit of choice is {}".format(kwargs['fruit']))
    else:
        print("No fruit found")
myfunc(fruit='apple',veggie='lettuce')
```

```python
def myfunc(*args,**kwargs):
    print(args)
    print(kwargs)
    print('I would like  {} {}'.format(args[0],kwargs['food']))
myfunc(10,20,30,fruit='apple',food='kiwi',animal='dog')
```

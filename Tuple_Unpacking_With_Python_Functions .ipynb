```python
wh=[("Pappu",100),("sara",200),("Ben",780)]
def ec(wh):
    current_max=0
    eom=''
    for employee,hours in wh:
        if hours>current_max:
            current_max=hours
            eom=employee
        else:
            pass
            
    return(eom,current_max)
r=ec(wh)
print(r)
```

```python
name,hours=ec(wh)       #tuple unpacking!!
print(name)
print(hours)
```

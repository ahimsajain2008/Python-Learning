```python
#check no is even or not using if-else statement
def even(n):
    if n%2==0:
        print("Even number!!")
    else:
        print("Odd number!!")
n=int(input("Enter the number:"))
even(n)
```

```python
#for series of number
def num(n):
    for i in range(0,n+1):
        if i%2==0:
            print(f'{i} is Even number!!')
        else:
            print(f'{i} is odd number!!')
num(10)
```

```python
#Return true if any number is even inside the list
def even(l1):
    for num in l1:
        if num%2==0:
            return True
        else:
            pass
```

```python
even([2,4,6])
```

```python
even([1,3,5,7,9]) #NO output because odd no and we used pass statement
```

```python
even([1,3,5,4]) #passed till element 5 and true at the element 4
```

```python
even([2,1,3,5])
```

```python
#Return true if any number is even inside the list
##Return false if any number is odd inside the list
def even(l1):
    for n in l1:
        if n%2==0:
            return True
        else:
            pass #dont use return false in this line it will check only 1 value in the list not the entire list
    return False
```

```python
even([1,3,5])
```

```python
even([2,3,4])
```

```python
#return the even no list if it is true
def even(l1):
    el=[]
    for n in l1:
        if n%2==0:
            el.append(n)
        else:
            pass
    return el
```

```python
even([1,2,3,4,5,6,7,8,9])
```

```python
even([1,3,5,7])
```
